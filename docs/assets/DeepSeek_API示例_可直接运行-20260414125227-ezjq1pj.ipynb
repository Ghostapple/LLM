```python
from openai import OpenAI

api_key = "sk-fd8f0ebb763c4af1912ddfb6a030fc60"

client = OpenAI(
    api_key=api_key,
    base_url="https://api.deepseek.com",
)

print("DeepSeek client initialized")
```

```python
def ask_deepseek(prompt: str, model: str = "deepseek-chat") -> str:
    response = client.chat.completions.create(
        model=model,
        messages=[
            {"role": "system", "content": "You are a concise, accurate, and friendly assistant."},
            {"role": "user", "content": prompt},
        ],
        temperature=0.7,
    )
    return response.choices[0].message.content


reply = ask_deepseek("什么是大语言模型？大语言模型有什么用？")
print(reply)
```

```python
def ask_deepseek_stream(prompt: str, model: str = "deepseek-chat") -> str:
    stream = client.chat.completions.create(
        model=model,
        messages=[
            {"role": "system", "content": "You are a concise, accurate, and friendly assistant."},
            {"role": "user", "content": prompt},
        ],
        temperature=0.7,
        stream=True,
    )

    chunks = []
    for event in stream:
        delta = event.choices[0].delta.content if event.choices[0].delta else None
        if delta:
            print(delta, end="", flush=True)
            chunks.append(delta)

    print()
    return "".join(chunks)
reply = ask_deepseek_stream("乌鸦为什么像写字台？")

# Uncomment the next line to test streaming output.
# ask_deepseek_stream("Explain large language models in three bullet points.")
```
